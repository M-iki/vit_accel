```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the Gaussian (Normal) distribution
mean = 0.0
std_dev = 1.0
num_samples = 10000
num_bins = 50  # number of bins

# Generate random samples
data = np.random.normal(loc=mean, scale=std_dev, size=num_samples)

# Compute histogram
counts, bin_edges = np.histogram(data, bins=num_bins, density=True)

# Compute bin centers
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

# Plot the bin centers as a continuous line
plt.figure(figsize=(8, 5))
plt.plot(bin_centers, counts, '-', color='blue', lw=2, label='Histogram (bin centers)')

# Overlay the ideal Gaussian curve for comparison
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
pdf = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean)/std_dev)**2)
plt.plot(x, pdf, 'r-', lw=2, label='True Gaussian PDF')

# Labels and legend
plt.title("Gaussian Distribution - Bin Centers")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True)
plt.show()
```
